```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('Indian_Traffic_Violations.csv')
df.head(5)
```

```python
df=pd.DataFrame(df)
```

```python
df['Helmet_Worn']=df['Helmet_Worn'].fillna('Unknown')
df['Seatbelt_Worn'] = df['Seatbelt_Worn'].fillna('Unknown')
```

```python
df.drop(columns=['Comments'],inplace=True)
```

### Which type of traffic violation occurs most frequently?

```python
df['Violation_Type'].value_counts().head(5)
```

### Which violation type results in the highest average fine amount?

```python
df.groupby('Violation_Type')['Fine_Amount'].mean().round(2)
```

### How many cases involve vehicles exceeding the speed limit?

```python
df['Over_Speeding'] = df['Recorded_Speed'] > df['Speed_Limit']
df['Over_Speeding'].value_counts()
```

### How do traffic violations vary by time of day (day vs night patterns)?

```python
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M')
df['Hour'] = df['Time'].dt.hour
df['Time_Period'] = df['Hour'].apply(lambda x: 'Day' if 6 <= x < 18 else 'Night')
df['Time_Period'].value_counts()
```

# Violation & Enforcement Patterns

### Which locations record the highest number of traffic violations?

```python
num = df['Location'].value_counts().head(10)

plt.bar(num.index, num.values)
plt.xticks(rotation=45)
plt.xlabel('Location')
plt.ylabel('Number of Violations')
plt.title('Top 10 Locations with Highest Traffic Violations')
plt.show()
```

### How do traffic violations vary across different vehicle types?

```python
count= df['Vehicle_Type'].value_counts()
plt.bar(count.index, count.values)
plt.xticks(rotation=45)
plt.xlabel('Vehicle Type')
plt.ylabel('Number of Violations')
plt.title('Traffic Violations as per Vehicle Type')
plt.show()
```

### Peak Violation Hours

```python
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M')
df['Hour'] = df['Time'].dt.hour
hour_counts = df['Hour'].value_counts().sort_index()
plt.figure(figsize=(10,5))
plt.plot(hour_counts.index, hour_counts.values, marker='o')
plt.xticks(range(0,24))
plt.xlabel('Hour of the Day')
plt.ylabel('Number of Violations')
plt.title('Peak Traffic Violation Hours')
plt.grid(True)
plt.show()
```

### Driver Age Distribution

```python
bins = [18,25,35,45,60,100]
labels = ['18-25','26-35','36-45','46-60','60+']
df['Age_Group'] = pd.cut(df['Driver_Age'], bins=bins, labels=labels)

age_counts = df['Age_Group'].value_counts().sort_index()
age_counts.plot(kind='bar', color='orange')
plt.xlabel('Age Group')
plt.ylabel('Number of Violations')
plt.title('Traffic Violations by Driver Age Group')
plt.show()
```

### Fine Paid vs Not Paid

```python
df['Fine_Paid'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Fine Payment Status')
plt.show()
```

### Over-speeding vs Within Speed Limit

```python
df['Over_Speeding'] = df['Recorded_Speed'] > df['Speed_Limit']
df[['Recorded_Speed', 'Speed_Limit', 'Over_Speeding']].head()
df['Over_Speeding'].value_counts()
df['Over_Speeding'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Over-speeding vs Within Speed Limit')
plt.ylabel('')
plt.show()

```

### Speed Difference → Box Plot

```python
df['Speed_Diff'] = df['Recorded_Speed'] - df['Speed_Limit']

df.boxplot(column='Speed_Diff', by='Vehicle_Type')
plt.title('Speed Difference by Vehicle Type')
plt.suptitle('')
plt.xlabel('Vehicle Type')
plt.ylabel('Speed Difference')
plt.show()
```

### Peak Hours → Heatmap

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Day'] = df['Date'].dt.day_name()

heat = pd.crosstab(df['Day'], df['Hour'])

plt.imshow(heat, aspect='auto')
plt.colorbar(label='Violations')
plt.xticks(range(24), range(24))
plt.yticks(range(len(heat.index)), heat.index)
plt.title('Violation Heatmap by Day and Hour')
plt.xlabel('Hour')
plt.ylabel('Day')
plt.show()
```

### Repeat Offender

```python
df['Repeat_Offender'] = df['Previous_Violations'] > 0

counts = df['Repeat_Offender'].value_counts()

plt.pie(counts, labels=['Repeat', 'First-time'],
        autopct='%1.1f%%', startangle=90)
centre = plt.Circle((0,0),0.70,fc='white')
plt.gca().add_artist(centre)
plt.title('Repeat vs First-time Offenders')
plt.show()
```

### Age Group × Repeat Offender 

```python
heat = pd.crosstab(df['Age_Group'], df['Repeat_Offender'])

plt.imshow(heat, aspect='auto')
plt.colorbar(label='Count')
plt.xticks([0,1], ['First-time','Repeat'])
plt.yticks(range(len(heat.index)), heat.index)
plt.title('Repeat Offenders by Age Group')
plt.xlabel('Offender Type')
plt.ylabel('Age Group')
plt.show()
```
